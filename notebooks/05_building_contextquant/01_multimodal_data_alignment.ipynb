# 01 — Multimodal Data Alignment

Align temporal market windows with same-day sentiment features.

```python
import numpy as np
import pandas as pd

np.random.seed(123)
```

```python
dates = pd.date_range('2024-01-01', periods=220, freq='B')
market = pd.DataFrame({
    'close': 100 + np.cumsum(np.random.normal(0.02, 1.0, len(dates))),
    'volume': np.random.randint(1_000_000, 5_000_000, len(dates))
}, index=dates)

news = pd.DataFrame({
    'date': dates,
    'sentiment_score': np.random.uniform(-1, 1, len(dates)),
    'sentiment_conf': np.random.uniform(0.5, 0.99, len(dates))
}).set_index('date')

merged = market.join(news, how='inner')
merged.head()
```

```python
window = 20
X_price, X_text, y = [], [], []

for i in range(window, len(merged) - 1):
    chunk = merged.iloc[i-window:i]
    X_price.append(chunk[['close', 'volume']].values)
    X_text.append(merged[['sentiment_score', 'sentiment_conf']].iloc[i].values)
    y.append(int(merged['close'].iloc[i+1] > merged['close'].iloc[i]))

X_price = np.array(X_price)
X_text = np.array(X_text)
y = np.array(y)

print('X_price shape:', X_price.shape)
print('X_text shape :', X_text.shape)
print('y shape      :', y.shape)
```
