# 03 — Evaluating Sequence Models

Evaluate directional models with confusion matrix and precision/recall.

```python
import numpy as np
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score

np.random.seed(99)
```

```python
y_true = np.random.randint(0, 2, 500)
# pretend this came from your trained LSTM
y_pred = np.where(np.random.rand(500) < 0.68, y_true, 1 - y_true)

acc = accuracy_score(y_true, y_pred)
cm = confusion_matrix(y_true, y_pred)
print('Accuracy:', round(acc, 4))
print('Confusion matrix:
', cm)
print('
Classification report:
', classification_report(y_true, y_pred, digits=3))
```

## Baseline comparison

```python
# Baseline: always predict majority class
majority = np.bincount(y_true).argmax()
baseline_pred = np.full_like(y_true, majority)
baseline_acc = accuracy_score(y_true, baseline_pred)

print('Model acc   :', round(acc, 4))
print('Baseline acc:', round(baseline_acc, 4))
print('Lift        :', round(acc - baseline_acc, 4))
```
