# 01 — Sequence Modeling Intro

Convert tabular time-series rows into 3D sequence tensors for recurrent models.

```python
import numpy as np
import pandas as pd

np.random.seed(1)
```

```python
n = 500
close = 100 + np.cumsum(np.random.normal(0.03, 1.0, n))
volume = np.random.randint(1_000_000, 4_000_000, n)
df = pd.DataFrame({'close': close, 'volume': volume})
df['ret'] = df['close'].pct_change().fillna(0.0)
df.head()
```

```python
features = df[['close', 'volume', 'ret']].values
target = (df['close'].shift(-1) > df['close']).astype(int).values[:-1]
features = features[:-1]

def make_windows(X, y, w=20):
    X_seq, y_seq = [], []
    for i in range(w, len(X)):
        X_seq.append(X[i-w:i])
        y_seq.append(y[i])
    return np.array(X_seq), np.array(y_seq)

X_seq, y_seq = make_windows(features, target, w=20)
print('X_seq shape:', X_seq.shape)  # [samples, time, features]
print('y_seq shape:', y_seq.shape)
```
