# 02 — Transformer Basics

Use a lightweight Transformer encoder for sentence classification.

```python
import torch
from torch import nn

torch.manual_seed(5)
```

```python
batch_size, seq_len, d_model = 16, 12, 32
X = torch.randn(batch_size, seq_len, d_model)
y = torch.randint(0, 3, (batch_size,))

encoder_layer = nn.TransformerEncoderLayer(
    d_model=d_model, nhead=4, dim_feedforward=64, batch_first=True
)
encoder = nn.TransformerEncoder(encoder_layer, num_layers=2)
head = nn.Linear(d_model, 3)

out = encoder(X)
cls_like = out[:, 0, :]
logits = head(cls_like)
print('logits shape:', logits.shape)
```

```python
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(list(encoder.parameters()) + list(head.parameters()), lr=1e-3)

for step in range(20):
    optimizer.zero_grad()
    logits = head(encoder(X)[:, 0, :])
    loss = criterion(logits, y)
    loss.backward()
    optimizer.step()

print('Final loss:', round(loss.item(), 4))
```
