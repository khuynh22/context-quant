# 03 — FinBERT Sentiment Pipeline

Build a practical sentiment inference pipeline with optional Hugging Face fallback.

```python
import pandas as pd

headlines = [
    'Company reports strong quarterly profit growth',
    'Regulatory concerns weigh on sector outlook',
    'Management raises full-year guidance',
    'Unexpected legal costs pressure earnings',
]
df = pd.DataFrame({'headline': headlines})
df
```

```python
def rule_based_sentiment(text):
    positive_words = {'strong', 'growth', 'raises', 'profit', 'beat'}
    negative_words = {'concerns', 'weigh', 'unexpected', 'legal', 'pressure', 'cut'}
    t = text.lower()
    p = sum(w in t for w in positive_words)
    n = sum(w in t for w in negative_words)
    if p > n:
        return 'positive', min(0.95, 0.55 + 0.15 * (p - n))
    if n > p:
        return 'negative', min(0.95, 0.55 + 0.15 * (n - p))
    return 'neutral', 0.5
```

```python
df[['label', 'confidence']] = df['headline'].apply(lambda t: pd.Series(rule_based_sentiment(t)))
df
```
