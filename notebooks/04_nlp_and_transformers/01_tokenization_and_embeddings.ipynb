# 01 — Tokenization and Embeddings

Turn raw financial text into vectors usable by neural networks.

```python
import re
from collections import Counter
import torch

texts = [
    'Company beats earnings expectations',
    'Guidance cut amid macro uncertainty',
    'Stock rallies after strong quarterly revenue',
]
```

```python
def tokenize(text):
    return re.findall(r"[a-zA-Z']+", text.lower())

tokens = [tok for t in texts for tok in tokenize(t)]
vocab = {'<pad>': 0, '<unk>': 1}
for w, _ in Counter(tokens).most_common():
    vocab[w] = len(vocab)

print('Vocab size:', len(vocab))
print(vocab)
```

```python
max_len = 8
encoded = []
for t in texts:
    idx = [vocab.get(tok, 1) for tok in tokenize(t)]
    idx = idx[:max_len] + [0] * max(0, max_len - len(idx))
    encoded.append(idx)

X = torch.tensor(encoded, dtype=torch.long)
embedding = torch.nn.Embedding(num_embeddings=len(vocab), embedding_dim=16, padding_idx=0)
emb = embedding(X)

pooled = emb.mean(dim=1)
print('Token tensor shape:', X.shape)
print('Embedding shape:', emb.shape)
print('Pooled shape:', pooled.shape)
```
