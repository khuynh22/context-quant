# 01 — Linear Layers and Activations

Build intuition for MLPs (multi-layer perceptrons) and activation functions.

```python
import torch
from torch import nn
import matplotlib.pyplot as plt

torch.manual_seed(21)
```

```python
x = torch.linspace(-5, 5, 200).unsqueeze(1)
relu = nn.ReLU()(x)
tanh = nn.Tanh()(x)
sigmoid = nn.Sigmoid()(x)

plt.figure(figsize=(10,4))
plt.plot(x.numpy(), relu.numpy(), label='ReLU')
plt.plot(x.numpy(), tanh.numpy(), label='Tanh')
plt.plot(x.numpy(), sigmoid.numpy(), label='Sigmoid')
plt.title('Activation Functions')
plt.legend(); plt.grid(alpha=0.3)
```

## Tiny MLP

```python
model = nn.Sequential(
    nn.Linear(2, 16),
    nn.ReLU(),
    nn.Linear(16, 16),
    nn.ReLU(),
    nn.Linear(16, 2)
)

sample = torch.randn(4, 2)
out = model(sample)
print('Input shape:', sample.shape)
print('Output shape:', out.shape)
```
