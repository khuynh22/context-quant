# 02 — Loss Functions and Optimizers

Compare common losses and optimizers for classification.

```python
import torch
from torch import nn
from torch.utils.data import TensorDataset, DataLoader

torch.manual_seed(33)
```

```python
X = torch.randn(1200, 10)
true_w = torch.randn(10, 3)
y = (X @ true_w).argmax(dim=1)
loader = DataLoader(TensorDataset(X, y), batch_size=64, shuffle=True)
```

```python
def train_once(optimizer_name='adam', epochs=12):
    model = nn.Sequential(nn.Linear(10, 32), nn.ReLU(), nn.Linear(32, 3))
    criterion = nn.CrossEntropyLoss()
    if optimizer_name == 'adam':
        optimizer = torch.optim.Adam(model.parameters(), lr=1e-2)
    else:
        optimizer = torch.optim.SGD(model.parameters(), lr=0.05, momentum=0.9)

    for _ in range(epochs):
        for xb, yb in loader:
            optimizer.zero_grad()
            logits = model(xb)
            loss = criterion(logits, yb)
            loss.backward()
            optimizer.step()

    with torch.no_grad():
        preds = model(X).argmax(dim=1)
        acc = (preds == y).float().mean().item()
    return acc

acc_adam = train_once('adam')
acc_sgd = train_once('sgd')
print('Adam accuracy:', round(acc_adam, 4))
print('SGD  accuracy:', round(acc_sgd, 4))
```
